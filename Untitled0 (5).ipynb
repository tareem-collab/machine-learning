```python

study_hours = [1, 2, 3, 4, 5, 6, 7]
marks = [50, 55, 60, 65, 72, 80, 88]


print("Number of samples:", len(study_hours))


print("Average study hours:", sum(study_hours) / len(study_hours))


print("Average marks:", sum(marks) / len(marks))


print("Highest mark:", max(marks))
print("Lowest mark:", min(marks))


print("\nEach sample:")

for i in range(len(study_hours)):
    print("Study hours:", study_hours[i], "Mark:", marks[i])


print("\nPass/Fail:")

for i in range(len(marks)):
    if marks[i] >= 60:
        print("Study hours:", study_hours[i],
              "Mark:", marks[i],
              "Pass")
    else:
        print("Study hours:", study_hours[i],
              "Mark:", marks[i],
              "Fail")

passes = 0

for mark in marks:
    if mark >= 60:
        passes += 1

print("\nNumber of students who passed:", passes)
```
