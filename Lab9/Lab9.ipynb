## Import Libraries

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## Get the Data

```python
data_path = Path('kyphosis.csv')
if data_path.is_file():
    df = pd.read_csv(data_path)
else:
    df = pd.read_csv('https://vincentarelbundock.github.io/Rdatasets/csv/rpart/kyphosis.csv')
df = df.drop(columns='rownames', errors='ignore')
```

```python
df.head()
```

## EDA

We'll just check out a simple pairplot for this small dataset.

```python
sns.pairplot(df,hue='Kyphosis',palette='Set1')
```

## Train Test Split

Let's split up the data into a training set and a test set!

```python
from sklearn.model_selection import train_test_split
```

```python
X = df.drop('Kyphosis',axis=1)
y = df['Kyphosis']
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.30, random_state=42, stratify=y
)
```

## Decision Trees

We'll start just by training a single decision tree.

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
dtree = DecisionTreeClassifier(random_state=42)
```

```python
dtree.fit(X_train,y_train)
```

## Prediction and Evaluation 

Let's evaluate our decision tree.

```python
predictions = dtree.predict(X_test)
```

```python
from sklearn.metrics import classification_report,confusion_matrix
```

```python
print(classification_report(y_test, predictions, zero_division=0))
```

```python
print(confusion_matrix(y_test,predictions))
```

## Tree Visualization
Plot the fitted decision tree with scikit-learn.

```python
from sklearn.tree import plot_tree
```

```python
plt.figure(figsize=(18, 9))
plot_tree(dtree, feature_names=X.columns.tolist(), class_names=dtree.classes_.tolist(),
          filled=True, rounded=True, fontsize=8)
plt.show()
```

## Random Forests

Now let's compare the decision tree model to a random forest.

```python
from sklearn.ensemble import RandomForestClassifier

rfc = RandomForestClassifier(n_estimators=100, random_state=42)
rfc.fit(X_train, y_train)
```

```python
rfc_pred = rfc.predict(X_test)
```

```python
print(confusion_matrix(y_test,rfc_pred))
```

```python
print(classification_report(y_test, rfc_pred, zero_division=0))
```
