```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('StudentsPerformance.csv')
df.head()
```

```python
print("Dataset shape:", df.shape)
print("\nData types:")
print(df.dtypes)
```

```python
print("Missing values:")
print(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())
```

```python
df.describe()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['math score'], bins=20, kde=True)
plt.title('Distribution of Math Scores')
plt.xlabel('Math Score')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x='gender', data=df)
plt.title('Gender Distribution')
plt.xlabel('Gender')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(x='gender', y='math score', data=df)
plt.title('Math Scores by Gender')
plt.xlabel('Gender')
plt.ylabel('Math Score')
plt.show()
```

Observation: Male students have a higher median math score than female students, although the score distributions overlap

```python
plt.figure(figsize=(7,5))
sns.scatterplot(x='reading score', y='writing score', data=df)
plt.title('Reading Score vs Writing Score')
plt.xlabel('Reading Score')
plt.ylabel('Writing Score')
plt.show()
```

Observation: There is a strong positive relationship between reading and writing scores. Students with higher reading scores tend to have higher writing scores.

```python
scores = df[['math score', 'reading score', 'writing score']]

plt.figure(figsize=(7,5))
sns.heatmap(scores.corr(), annot=True)
plt.title('Correlation Between Scores')
plt.show()
```

Observation: All three scores are positively correlated. Reading and writing scores have the strongest correlation (0.95), while math is also strongly correlated with reading (0.82) and writing (0.80).

```python
plt.figure(figsize=(7,5))
sns.boxplot(x='test preparation course', y='math score', data=df)
plt.title('Math Scores by Test Preparation Course')
plt.xlabel('Test Preparation Course')
plt.ylabel('Math Score')
plt.show()
```

Observation: Students who completed the test preparation course generally achieved higher math scores than students who did not complete the course.

```python
average_scores = df[['math score', 'reading score', 'writing score']].mean()
print(average_scores)
```

Observation: The average reading score is the highest (69.17), followed by writing (68.05) and math (66.09).

Conclusion

The dataset contains 1000 records and 8 features, with no missing values or duplicate rows. The EDA showed a strong positive relationship between reading and writing scores, and all three exam scores were positively correlated. Students who completed the test preparation course generally achieved higher math scores. Overall, the analysis helped identify important patterns and relationships in students’ academic performance.
